```python
import csv

data = [
    ["Match", "Team1", "Team2", "Winner"],
    ["Match 1", "CSK", "MI", "CSK"],
    ["Match 2", "RCB", "KKR", "KKR"],
    ["Match 3", "GT", "RR", "GT"],
    ["Match 4", "MI", "DC", "MI"],
    ["Match 5", "CSK", "RCB", "RCB"]
]

file = open("ipl_matches.csv", "w", newline="")

writer = csv.writer(file)
writer.writerows(data)

file.close()

print("CSV file created successfully!")
```
