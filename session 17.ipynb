```python
from datetime import datetime, timedelta
import os

recharge_date = datetime.strptime("05/11/2025", "%d/%m/%Y")

validities = [14, 28, 56, 84, 90]

os.makedirs("recharge_out", exist_ok=True)

results = []

for days in validities:
    expiry_date = recharge_date + timedelta(days=days)
    expiry = expiry_date.strftime("%d/%m/%Y")

    result = f"{days}-day plan expiry: {expiry}"
    print(result)
    results.append(result)

with open("recharge_out/expiry_dates.txt", "w") as file:
    for result in results:
        file.write(result + "\n")
```
