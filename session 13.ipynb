```python
def name(hello):
    reverse = hello[::-1]
    return reverse
print(name("hello"))
```

```python
def sum_playlist_duration(playlist):
    if len(playlist) == 0:
        return 0
    return playlist[0] + sum_playlist_duration(playlist[1:])
duration = [120,180,240,300]

print(sum_playlist_duration(duration))
```

```python
def count_like(post):
    total = post["likes"]

    for i in post["replies"]:
        total = total + count_like(i)
    return total
post = {"likes":8,"replies":[{"likes":9,"replies":[]},{"likes":7,"replies":[]}]}
print(count_like(post))
```

```python
app_status = "offline"
def whatsapp():
    user_status = "online"
    print("inside function:",user_status)
    print("app_status:",user_status)
print("before function:",app_status)
whatsapp()
print("after function:",app_status)
```
