```python
def clean_brand_name(name):
    return name.strip().replace("-", " ")

print(clean_brand_name(" oneplus-Nord "))
```

# task 3

```python
name = "Apple iPhone 14 Pro Max"

split_point = len(name.split()[0])  # Length of the first word: "Apple"

brand = name[:split_point]          # Slice from the start to "Apple"
model = name[split_point:].strip()  # Slice the rest and remove the starting space

print("Brand:", brand)
print("Model:", model)
```

# task 4

```python
def format_product_display(name, price):
    return f"{name} - ₹{price}"

print(format_product_display("Boat Earbuds", 1299))
```
