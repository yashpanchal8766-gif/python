```python
def insta_posts_generater(posts):
      for position, post in enumerate(posts, start=1):
        yield post, position

nupost = [  "Good morning! ☀️","Enjoyi my day! 😎","New photo 📸","Keep smiling 😊"]
post_generater = insta_posts_generater(nupost)

while True:
    try:
        print(next(post_generater))
    except StopIteration:
        break
```

```python
def cashback_generator(transactions):
    for amount in transactions:
        yield amount * 0.05


transactions = [100, 500, 1000, 200, 800]

cashback = cashback_generator(transactions)

for value in cashback:
    print(value)
```
