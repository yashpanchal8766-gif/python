```python
gst_price = lambda price: price+(price * 18/100)

print(gst_price(100))
print(gst_price(250))
print(gst_price(500))
```

```python
songs = [' shape OF you ', ' blinding LIGHTS ', ' perfect ', ' let ME down slowly ']
clean_songs = list(map(lambda song:song.strip().title(), songs))
print(clean_songs)
```

```python
products = ["Samsung Phone", "Laptop", "Shoes", "Smart Watch", "Mouse", "Speaker"]

s_products = list(filter(lambda product : product.startswith("S"), products))
print(s_products)
```

```python
from functools import reduce

orders = [120,340,540,80]
total_order = reduce(lambda x,y:x+y,orders)
print(total_order)
```

```python
from functools import reduce
numbers = [40,60,80,120]

double_numbers = list(map(lambda x:x*2,numbers))
filtered_numbers = list(filter(lambda x:x>100,double_numbers))
total_numbers = reduce(lambda x,y:x+y,filtered_numbers)
print(total_numbers)
```
